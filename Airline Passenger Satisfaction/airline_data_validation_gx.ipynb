## Data Validation

### Inisialisasi Data Context

```python
# Menginisialisasi data context di folder saat ini
context = FileDataContext.create(project_root_dir='./')
```

### Connect ke Datasource

```python
# Mendefinisikan nama Datasource
datasource_name = 'airlines'
# Menambahkan datasource berbasis pandas ke dalam Data Context
datasource = context.sources.add_pandas(datasource_name)

# Mendefinisikan nama dataset
asset_name = 'flight-satisfaction'
# Menentukan path ke file dataset
path_to_data = 'train.csv' # Pastikan file train.csv ada di folder yang sama
# Menambahkan file CSV sebagai data asset ke dalam datasource
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)

# Membangun batch request dari dataset
batch_request = asset.build_batch_request()
```

### Membuat Expectation Suite

```python
# Mendefinisikan nama expectation suite
expectation_suite_name = 'flight-expectation-suite'
# Membuat atau update expectation suite di dalam data context
context.add_or_update_expectation_suite(expectation_suite_name)

# Membuat validator untuk menghubungkan data dengan expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name)

# Menampilkan preview data melalui validator
validator.head()
```

#### Ekspektasi 1: kolom `id` harus unique

Pada tahap ini, dilakukan pengecekan terhadap kolom `id`. Hal ini bertujuan untuk memastikan bahwa setiap baris data mewakili satu entitas penumpang yang berbeda dan tidak terdapat duplikasi pencatatan dalam sistem

```python
validator.expect_column_values_to_be_unique('id')
```

#### Ekspektasi 2: Semua rating harus antara 0 sampai 5

Pada tahap ini, dilakukan validasi terhadap seluruh kolom yang merepresentasikan penilaian kepuasan penumpang (rating). Skala penilaian yang digunakan dalam survei ini adalah **0 hingga 5**. Proses ini menggunakan perulangan (*looping*) untuk memastikan bahwa tidak ada nilai di luar rentang tersebut pada 14 kolom fasilitas maskapai.

```python
# Daftar kolom yang berisi data rating tingkat kepuasan (skala 0-5)
rating_cols = [
    'Inflight wifi service',
    'Departure/Arrival time convenient',
    'Ease of Online booking',
    'Gate location',
    'Food and drink',
    'Online boarding',
    'Seat comfort',
    'Inflight entertainment',
    'On-board service',
    'Leg room service',
    'Baggage handling',
    'Checkin service',
    'Inflight service',
    'Cleanliness']

# Melakukan iterasi untuk setiap kolom dalam daftar rating_cols
for col in rating_cols:
    # Memastikan nilai pada setiap kolom rating berada di rentang min 0 dan max 5
    validator.expect_column_values_to_be_between(col, min_value=0, max_value=5)
```

#### Ekspektasi 3: kolom `Class` hanya terdiri dari kategori "Business", "Eco", "Eco Plus"

Validasi ini diterapkan pada kolom `Class` untuk memastikan bahwa data kategori kelas penerbangan hanya berisi nilai yang telah ditentukan, yaitu **Business**, **Eco**, atau **Eco Plus**. Hal ini penting untuk menjaga konsistensi data kategorikal agar tidak ada nilai di luar standar yang masuk ke dalam sistem.

```python
validator.expect_column_values_to_be_in_set('Class', ["Business", "Eco", "Eco Plus"])
```

#### Ekspektasi 4: kolom `Flight Distance` harus bertipe integer

Validasi ini bertujuan untuk memastikan konsistensi tipe data pada kolom `Flight Distance`. Pengecekan dilakukan untuk menjamin bahwa data yang masuk bertipe numerik, sehingga tidak terjadi kesalahan saat dilakukan perhitungan matematis atau analisis statistik pada tahap berikutnya

```python
validator.expect_column_values_to_be_in_type_list('Flight Distance', ['int64', 'int'])
```

#### Ekspektasi 5: kolom `Gender` harus terdiri dari 2 kategori

Validasi ini dilakukan pada kolom `Gender` untuk memastikan hanya memiliki tepat **2 nilai unik** (Male dan Female). Pengecekan ini berfungsi untuk mendeteksi nilai kosong yang tidak terdeteksi atau kesalahan input kategori baru yang tidak sesuai dengan struktur dataset.

```python
validator.expect_column_unique_value_count_to_be_between(
    column='Gender', # Kolom yang akan dicek jumlah uniknya
    min_value=2,     # Batas minimum nilai unik
    max_value=2)     # Batas maksimum nilai unik
```

#### Ekspektasi 6: Nilai maksimum Delay harus dalam batas wajar operasional


Validasi ini dilakukan pada kolom `Departure Delay in Minutes` dan `Arrival Delay in Minutes` untuk memastikan bahwa nilai keterlambatan maksimum yang tercatat masih berada dalam batas wajar operasional penerbangan (0 hingga 2000 menit). Hal ini berfungsi untuk mendeteksi adanya anomali data atau *outlier* ekstrem yang tidak masuk akal

```python
# Memastikan nilai maksimum Departure Delay masih dalam batas wajar operasional
validator.expect_column_max_to_be_between(
    'Departure Delay in Minutes',
    min_value=0,
    max_value=2000)

# Memastikan nilai maksimum Arrival Delay masih dalam batas wajar operasional
validator.expect_column_max_to_be_between(
    'Arrival Delay in Minutes',
    min_value=0,
    max_value=2000)
```

#### Ekspektasi 7: kolom `Age` tidak boleh negatif

Pada tahap ini, dilakukan validasi terhadap kolom `Age` untuk memastikan bahwa data umur penumpang memiliki nilai yang logis. Validasi ini menetapkan batas minimum **0 tahun**, guna memastikan tidak ada nilai negatif yang masuk akibat kesalahan input data saat proses registrasi penumpang

```python
# Age tidak boleh negatif
validator.expect_column_values_to_be_between('Age',min_value=0)
```

#### Save Expectation Suite

```python
validator.save_expectation_suite(discard_failed_expectations=False)
```
