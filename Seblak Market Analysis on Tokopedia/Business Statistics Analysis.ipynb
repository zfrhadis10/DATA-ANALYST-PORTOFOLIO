### Analisis 1: Statistik Deskriptif (Mengenal Karakteristik Data)

Langkah pertama sebelum mengambil keputusan bisnis adalah memahami seperti apa data yang kita miliki. Rosie ingin mengetahui gambaran umum tiga variabel kunci: **harga produk**, **banyaknya produk terjual**, dan **rating produk**. Ia menghitung rata-rata, median, standar deviasi, skewness, dan kurtosis untuk memahami distribusi dan mendeteksi adanya outlier pada data.

```python
# Membuat dictionary kolom yang akan dianalisis
kolom_analisis = {
    'harga produk': 'Harga Produk',
    'banyaknya terjual': 'Banyak Produk Terjual',
    'rating produk': 'Rating Produk'
}

# Membuat list kosong untuk menampung semua hasil perhitungan dari looping
hasil = []

# Looping untuk mencari Mean, Median,  Standar Deviasi, Skewness, dan Kurtosis pada masing-masing kolom
for kolom, label in kolom_analisis.items():
    data = df[kolom]
    hasil.append({
        'Variabel': label,
        'Mean': round(data.mean(), 2),
        'Median': round(data.median(), 2),
        'Std Dev': round(data.std(), 2),
        'Skewness': round(data.skew(), 4),
        'Kurtosis': round(data.kurt(), 4)
    })

# Mengubah List menjadi df
df_desc = pd.DataFrame(hasil).set_index('Variabel')

# Menampilkan df
df_desc
```

**Banyak Produk Terjual:**
- Rata-rata terjual **175,5 unit/bulan** dengan median hanya **70 unit**. Skewness **+4.00** dan kurtosis **16.29** menunjukkan distribusi **sangat condong ke kanan** dengan outlier ekstrem. Ada beberapa produk yang terjual sangat banyak (> 1.000 unit/bulan) sehingga menarik mean ke atas.
- Outlier ini bersifat alami karena adanya produk yang dijual satuan.

**Rating Produk:**

### Analisis 2: Confidence Interval Pendapatan (Potensi pendapatan Dropship)

Setelah memahami karakteristik data, Rosie ingin menjawab pertanyaan yang paling krusial sebelum memulai bisnis: **"Berapa kisaran pendapatan bulanan yang bisa ia harapkan?"** Ia menghitung pendapatan per produk (harga x unit terjual) lalu menggunakan **Confidence Interval 95%** untuk mendapatkan estimasi rentang pendapatan dengan asumsi data terdistribusi normal dan data terjual merupakan penjualan per bulan.

```python
# Menghitung pendapatan per produk
df['pendapatan'] = df['harga produk'] * df['banyaknya terjual']

# Menghitung standar deviasi
std_pendapatan = df['pendapatan'].std()
mean_pendapatan = df['pendapatan'].mean()

# Menghitung
n = len(df['pendapatan'])
low, up = stats.norm.interval(0.95,loc=mean_pendapatan,scale=std_pendapatan/np.sqrt(n))
print('Lower Limit:',low)
print('Upper Limit:',up)
```

Dengan tingkat kepercayaan **95%**, rata-rata pendapatan
populasi produk seblak per bulan berada pada rentang:

**Rp 1,284,356** s.d. **Rp 7,635,361**

**Insight Analisis 2**

Dengan **confidence interval 95%**, estimasi rata-rata pendapatan bulanan per produk seblak yang dijual di Tokopedia berada pada rentang **Rp 1.284.355,87 hingga Rp 7.635.361,46**.

Artinya:
- Dalam skenario **minimum (lower bound)**: dropshipper bisa mengharapkan pendapatan sekitar **Rp 1,28 juta/bulan** per produk yang dijual
- Dalam skenario **maksimum (upper bound)**: pendapatan bisa mencapai **Rp 7,63 juta/bulan** per produk
- Rentang yang lebar ini mencerminkan **variasi lebar** antar produk. Ada produk dengan harga satuan murah (Indomie Seblak ~Rp 3.500) dengan volume besar, dan ada produk grosir (Indomie karton isi 40 ~Rp 153.000) dengan harga tinggi namun volume lebih kecil.


### Analisis 3: Uji Hipotesis. Apakah Harga di Jabodetabek Berbeda dari Luar Jabodetabek?


Rosie tinggal di area Jabodetabek dan penasaran apakah terdapat perbedaan harga dengan penjual dari luar Jabodetabek. Biaya bahan baku, logistik, dan sewa di Jabodetabek umumnya lebih tinggi. Apakah ini mempengaruhi harga jual pada produk seblak?

**Hipotesis:**
- **H0:** Tidak ada perbedaan rata-rata harga produk seblak antara penjual di Jabodetabek dan di luar Jabodetabek
- **H1:** Terdapat perbedaan rata-rata harga produk seblak antara penjual di Jabodetabek dan di luar Jabodetabek

**Jenis uji:** Two-tailed, karena untuk mengetahui perbedaan antara wilayah Jabodetabek dan luar Jabodetabek (belum memiliki asumsi)

**Tingkat signifikansi:** α = 0.05

```python
# Membuat daftar nama kota yang termasuk kategori Jabodetabek
kota_jabodetabek = [
    'Jakarta Selatan', 'Jakarta Timur', 'Jakarta Barat',
    'Jakarta Utara', 'Jakarta Pusat', 'Depok', 'Bekasi',
    'Tangerang', 'Tangerang Selatan', 'Bogor', 'Kab. Bekasi']

 # Membuat kolom baru 'wilayah' menggunakan fungsi lambda untuk memberi label pada data
df['wilayah'] = df['kota toko'].apply(lambda x: 'Jabodetabek' if x in kota_jabodetabek else 'Luar Jabodetabek')

# Memisahkan data harga ke dalam dua variabel berbeda untuk dibandingkan
jabod = df[df['wilayah'] == 'Jabodetabek']['harga produk']
luar  = df[df['wilayah'] == 'Luar Jabodetabek']['harga produk']

# Menghitung jumlah produk dan median pada dua variabel
print(f'Jumlah produk Jabodetabek: {len(jabod)} produk')
print(f'Median harga Jabodetabek: Rp {jabod.median():,.2f}') # Menggunakan median karena terdapat outlier
print(f'Jumlah produk Luar Jabodetabek: {len(luar)} produk')
print(f'Median harga Luar Jabodetabek: Rp {luar.median():,.2f}') # Menggunakan median karena terdapat outlier
```

```python
# Uji Normalitas pada data menggunakan Shapiro-Wilk (agar bisa menentukan metode uji hipotesis yang digunakan)
stat_j, p_j = stats.shapiro(jabod)
stat_l, p_l = stats.shapiro(luar)
print(f'Jabodetabek : stat={stat_j:.4f}, p-value={p_j:.4f} → {"Normal" if p_j > 0.05 else "Tidak Normal"}')
print(f'Luar Jabod  : stat={stat_l:.4f}, p-value={p_l:.4f} → {"Normal" if p_l > 0.05 else "Tidak Normal"}')
```

Kedua variabel menunjukkan distribusi yang **tidak normal**. Sehingga untuk uji hipotesis menggunakan metode **Mann-Whitney U Test** (pengganti metode T-test)

```python
# Menjalankan uji Mann-Whitney U Test
stat_test, p_val = stats.mannwhitneyu(jabod, luar, alternative='two-sided')

# Menampilkan  hasil statistik
print(f'Statistic = {stat_test:.4f}')
print(f'p-value   = {p_val:.4f}')
```

**Kesimpulan**: Karena p-value (0.9928) > 0.05 maka menerima H0 → **Tidak ada perbedaan signifikan**

**Insight Analisis 3**

Rosie bebas memilih supplier dari wilayah mana saja tanpa perlu khawatir modalnya membengkak karena lokasi, karena **tidak ada perbedaan signifikan antara produk yang berasal dari Jabodetabek dan Luar Jabodetabek**. Ia bisa lebih fokus mencari supplier yang memiliki **pengiriman paling cepat** atau **biaya ongkir termurah** ke pelanggan daripada pusing memikirkan perbedaan harga produk antar wilayah.

### Analisis 4: Uji Korelasi. Apakah Konsumen Lebih Suka Produk Murah?


Setelah Rosie menyadari bahwa lokasi supplier tidak jadi masalah, ia kini menghadapi pertanyaan paling krusial bagi setiap pedagang: **"Seberapa sensitifkah pembeli seblak terhadap harga?"**
Rosie bimbang. Jika ia mengambil keuntungan (margin) terlalu besar, ia takut seblaknya tidak dilirik. Namun, jika terlalu murah, ia khawatir profitnya tidak cukup untuk biaya operasional (listrik, internet). Sebagai data analyst, Rosie tidak mau menebak-nebak. Ia pun melakukan **Uji Korelasi Spearman** untuk melihat hubungan antara harga dengan jumlah produk terjual.

**Hipotesis**:

**H0** : Harga produk tidak mempengaruhi banyaknya produk terjual

**H1** : Harga produk mempengaruhi banyaknya produk terjual

```python
# Menjalankan Uji Korelasi Spearman
corr_s, p_s = stats.spearmanr(df['harga produk'], df['banyaknya terjual'])

print(f'Koefisien Korelasi (r) = {corr_s:.4f}')
print(f'p-value                = {p_s:.4f}')
```

**Kesimpulan:**

Karena p-value (0.9825) > 0.05 maka:

*   Korelasi TIDAK SIGNIFIKAN → Tolak H0 dan menerima H1
*   **Tidak ada hubungan** linear yang bermakna antara harga dan rating



**Insight Analisis 4**

Karena p-value > 0.05, Rosie menyimpulkan bahwa **harga produk tidak memiliki pengaruh yang signifikan terhadap jumlah penjualan** seblak di Tokopedia. Artinya, pembeli seblak tidak semata-mata mencari harga yang paling murah.

Rosie menyadari bahwa ia tidak perlu terpaku pada strategi "jual yang paling murah untuk agar banyak yang beli" yang bisa merugikan margin keuntungannya. Daripada menjual dengan harga murah, Rosie memutuskan untuk menetapkan harga yang wajar (kompetitif namun tetap menguntungkan) dan akan lebih fokus pada faktor lain yang lebih diperhatikan konsumen, seperti kualitas foto produk, kecepatan respon chat, rating produk, karena tidak ada bukti statistik kalau harga murah bisa membuat produk menjadi laris.
